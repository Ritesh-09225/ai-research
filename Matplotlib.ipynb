##What is a Line Chart?
A line chart (or line plot) is a type of chart that displays information as a series of data points called 'markers' connected by straight line segments. It's primarily used to visualize data points over a continuous interval or time period.

##Why are Line Charts Used?
Line charts are excellent for:

Showing Trends Over Time: They clearly illustrate how a variable changes over a specific duration.<br>

Comparing Multiple Series: You can plot several lines on the same chart to compare trends between different categories or groups.

Identifying Patterns and Fluctuations: They make it easy to spot upward or downward trends, seasonality, cycles, and sudden changes in data.

Representing Continuous Data: Ideal for data that changes gradually rather than discretely.

Real-World Uses
Line charts are ubiquitous in various fields:

Finance: Tracking stock prices, currency exchange rates, or investment growth over time.

Healthcare: Monitoring a patient's vital signs (e.g., heart rate, temperature) or disease prevalence over months.

Business: Analyzing sales performance, website traffic, or customer engagement over weeks or quarters.

Meteorology: Displaying temperature fluctuations, rainfall patterns, or atmospheric pressure changes throughout the day or year.

Sports: Charting an athlete's performance metrics (e.g., speed, points scored) over a season.

##Use in Machine Learning
In Machine Learning, line charts are frequently used for:

Model Performance Over Epochs/Iterations: Visualizing how a model's loss or accuracy changes during training (e.g., learning curves).

Time Series Forecasting: Plotting actual vs. predicted values for time-dependent data (e.g., stock predictions, weather forecasts).

Hyperparameter Tuning: Showing the impact of different hyperparameter values on model metrics.

Feature Importance: While bar charts are often used, line charts can show how a feature's importance changes over time or with different model configurations.

Anomaly Detection: Highlighting deviations from an expected trend in time series data.

They provide an intuitive way to understand the dynamic behavior of data and model performance.

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data
x = np.linspace(0, 10, 100) # 100 points from 0 to 10
y = np.sin(x) # Sine wave

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(x, y, label='Sine Wave', color='blue', linestyle='-', marker='o', markersize=4)

# Add title and labels
plt.title('Sample Line Chart: Sine Wave Over Time')
plt.xlabel('Time (Units)')
plt.ylabel('Amplitude')

# Add a grid for better readability
plt.grid(True, linestyle='--', alpha=0.6)

# Add a legend
plt.legend()
# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data
x = np.linspace(0, 10, 100) # 100 points from 0 to 10
y1 = np.sin(x) # First sine wave
y2 = np.cos(x) # Second sine wave (cosine for variety)

# Create the plot
plt.figure(figsize=(10, 6))

# Plot the first line
plt.plot(x, y1, label='Sine Wave', color='blue', linestyle='-', marker='o', markersize=4)

# Plot the second line with different styles
plt.plot(x, y2, label='Cosine Wave', color='red', linestyle='--', marker='x', markersize=4)

# Add title and labels
plt.title('Multiple Line Chart: Sine and Cosine Waves with Annotations')
plt.xlabel('Time (Units)')
plt.ylabel('Amplitude')

# Add a grid for better readability
plt.grid(True, linestyle='--', alpha=0.6)

# Add a legend to distinguish the lines
plt.legend()

# Add annotations
# Annotate a peak on the Sine Wave
plt.annotate('Sine Peak', xy=(np.pi/2, 1), xytext=(np.pi/2 + 2, 0.8), # adjust xytext for annotation text position
             arrowprops=dict(facecolor='blue', shrink=0.05), # arrow properties
             fontsize=10, color='blue')

# Annotate a trough on the Cosine Wave
plt.annotate('Cosine Trough', xy=(np.pi, -1), xytext=(np.pi + 1, -0.8), # adjust xytext
             arrowprops=dict(facecolor='black', shrink=0.08),
             fontsize=10, color='red')

# Show the plot
plt.show()
```

##Subplots

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data
x = np.linspace(0, 10, 100) # 100 points from 0 to 10
y1 = np.sin(x) # First sine wave
y2 = np.cos(x) # Second sine wave (cosine for variety)

# Create a figure and a set of subplots
fig, axes = plt.subplots(2, 1, figsize=(10, 8)) # 2 rows, 1 column

# Plot the first line (Sine Wave) on the first subplot
axes[0].plot(x, y1, label='Sine Wave', color='blue', linestyle='-', marker='o', markersize=4)
axes[0].set_title('Sine Wave')
axes[0].set_xlabel('Time (Units)')
axes[0].set_ylabel('Amplitude')
axes[0].grid(True, linestyle='--', alpha=0.6)
axes[0].legend()

# Plot the second line (Cosine Wave) on the second subplot
axes[1].plot(x, y2, label='Cosine Wave', color='red', linestyle='--', marker='x', markersize=4)
axes[1].set_title('Cosine Wave')
axes[1].set_xlabel('Time (Units)')
axes[1].set_ylabel('Amplitude')
axes[1].grid(True, linestyle='--', alpha=0.6)
axes[1].legend()

# Adjust layout to prevent overlap
plt.tight_layout()

# Show the plot
plt.show()
```

Bar Chart

## What is a Bar Chart?
A bar chart (or bar graph) is a chart or graph that presents categorical data with rectangular bars, with heights or lengths proportional to the values that they represent. The bars can be plotted vertically or horizontally.

## Why are Bar Charts Used?
Bar charts are excellent for:

*   **Comparing different categories:** They make it easy to compare quantities of different categories at a glance.
*   **Showing changes over time (for discrete intervals):** While line charts are great for continuous time, bar charts can effectively show values for distinct time periods (e.g., monthly sales).
*   **Displaying distributions:** They can show how a single categorical variable is distributed.

## Real-World Uses
Bar charts are widely used in many fields:

*   **Business:** Comparing sales performance of different products, market share of competitors, or departmental budgets.
*   **Surveys:** Presenting the results of surveys, like satisfaction levels across different options.
*   **Demographics:** Showing population distribution by age group, education level, or geographic region.
*   **Science:** Comparing experimental results across different conditions.

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.patches as mpatches # Import for custom legend

# Sample data: Categories and their corresponding values
categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 56, 12, 39]

# Define a list of colors for each bar
bar_colors = ['#FF9999', '#66B2FF', '#99FF99', '#FFCC66', '#B266FF'] # Example colors

# Create the bar chart
plt.figure(figsize=(10, 6)) # Increased figure size for better legend display
bars = plt.bar(categories, values, color=bar_colors) # Pass the list of colors

# Add data labels to the top of each bar
#for bar in bars:
    #yval = bar.get_height()
    #plt.text(bar.get_x() + bar.get_width()/2, yval + 0.5, yval, ha='center', va='bottom') # x, y, text, horizontal alignment, vertical alignment

# Add title and labels
plt.title('Custom Bar Chart: Values per Category with Different Colors and Legend')
plt.xlabel('Category')
plt.ylabel('Value')

# Add grid for better readability (optional)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Create custom legend handles and labels
legend_patches = []
for i, category in enumerate(categories):
    legend_patches.append(mpatches.Patch(color=bar_colors[i], label=category))

# Add the custom legend
plt.legend(handles=legend_patches, title='Categories', bbox_to_anchor=(1.05, 1), loc='upper left')

# Adjust layout to make space for the legend outside the plot area
plt.tight_layout()

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.patches as mpatches

# Sample data: Categories and their corresponding values
categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 56, 12, 39]

# Define a list of colors for each bar
bar_colors = ['#FF9999', '#66B2FF', '#99FF99', '#FFCC66', '#B266FF'] # Example colors

# Create the horizontal bar chart
plt.figure(figsize=(10, 6))
# Use plt.barh for horizontal bars
bars = plt.barh(categories, values, color=bar_colors)

# Add data labels to the end of each bar
for bar in bars:
    xval = bar.get_width() # For horizontal bars, width is the value
    plt.text(xval + 0.5, bar.get_y() + bar.get_height()/2, xval, ha='left', va='center') # x, y, text, horizontal alignment, vertical alignment

# Add title and labels
plt.title('Horizontal Bar Chart: Values per Category')
plt.xlabel('Value')
plt.ylabel('Category')

# Add grid for better readability (optional)
plt.grid(axis='x', linestyle='--', alpha=0.7)

# Create custom legend handles and labels
legend_patches = []
for i, category in enumerate(categories):
    legend_patches.append(mpatches.Patch(color=bar_colors[i], label=category))

# Add the custom legend
plt.legend(handles=legend_patches, title='Categories', bbox_to_anchor=(1.05, 1), loc='upper left')

# Adjust layout to make space for the legend outside the plot area
plt.tight_layout()

# Show the plot
plt.show()
```

Pie Chart

## What is a Pie Chart?
A pie chart is a circular statistical graphic, which is divided into slices to illustrate numerical proportion. In a pie chart, the arc length of each slice (and consequently its central angle and area) is proportional to the quantity it represents.

## Why are Pie Charts Used?
Pie charts are commonly used for:

*   **Displaying proportions:** They are effective for showing how a whole is divided into different parts.
*   **Comparing parts of a whole:** They allow for a quick visual comparison of the relative sizes of different categories within a single dataset.

## Real-World Uses
Pie charts are used in various contexts:

*   **Business:** To show market share distribution among competitors, breakdown of expenses, or revenue by product category.
*   **Demographics:** To illustrate population distribution by age group, ethnicity, or gender.
*   **Surveys:** To present responses to multiple-choice questions as proportions of the total.
*   **Education:** To represent the allocation of a budget or the distribution of grades in a class.

```python
import matplotlib.pyplot as plt

# Sample data: Categories and their corresponding sizes
labels = ['Category A', 'Category B', 'Category C', 'Category D']
sizes = [15, 30, 45, 10] # Proportions for each category
colors = ['#FF9999', '#66B2FF', '#99FF99', '#FFCC66'] # Colors for each slice
explode = (0, 0.1, 0, 0)  # 'explode' the 2nd slice (Category B) a little

# Create the pie chart
plt.figure(figsize=(8, 8))
plt.pie(sizes, explode=explode, labels=labels, colors=colors,
        autopct='%1.1f%%', shadow=True, startangle=140)

plt.title('Distribution of Categories')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.

plt.show()
```

## What is a Scatter Plot?
A scatter plot (also known as a scatter graph or scatter diagram) is a type of plot or mathematical diagram using Cartesian coordinates to display values for typically two variables for a set of data. The data are displayed as a collection of points, each having the value of one variable determining the position on the horizontal axis and the value of the other variable determining the position on the vertical axis.

## Why are Scatter Plots Used?
Scatter plots are excellent for:

*   **Showing the relationship between two variables:** They allow for the visual identification of patterns and trends, such as positive or negative correlations, or no correlation at all.
*   **Identifying outliers:** Points that lie far away from the general cluster of points can be easily spotted as outliers.
*   **Visualizing data distribution:** They can give a sense of the spread or clustering of data points.

## Real-World Uses
Scatter plots are widely used in various fields:

*   **Statistics/Research:** To study the relationship between two quantitative variables, like height and weight, or study hours and exam scores.
*   **Economics:** To analyze the relationship between advertising spend and sales revenue, or interest rates and inflation.
*   **Healthcare:** To observe the correlation between drug dosage and patient response, or age and blood pressure.
*   **Machine Learning:** To visualize the relationship between features in a dataset, or to plot actual vs. predicted values for regression models to assess performance.

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate sample data
np.random.seed(42)
x_data = np.random.rand(100) * 10 # 100 random values between 0 and 10
y_data = 2 * x_data + np.random.randn(100) * 2 + 5 # Linear relationship with some noise

# Create the scatter plot using Matplotlib
plt.figure(figsize=(10, 6))
plt.scatter(x_data, y_data, c=y_data, cmap='viridis', s=x_data*10, alpha=0.7)

# Add title and labels
plt.title('Sample Scatter Plot: Relationship Between Two Variables (Matplotlib)')
plt.xlabel('X-Axis Data')
plt.ylabel('Y-Axis Data')

# Add grid for better readability (optional)
plt.grid(True, linestyle='--', alpha=0.6)

# Add a color bar to indicate the meaning of the color (optional)
cbar = plt.colorbar()
cbar.set_label('X-Axis Data Value')

# Show the plot
plt.show()
```

plt.scatter(...): This is the core function call from Matplotlib's pyplot module (aliased as plt) to create a scatter plot.

x_data, y_data: These are the primary data points for the scatter plot. x_data provides the coordinates for the horizontal (x) axis, and y_data provides the coordinates for the vertical (y) axis. Each pair (x_data[i], y_data[i]) forms a single point on the plot.

c=x_data: This argument specifies the color of each point. By setting c to x_data, the color of each point will correspond to its value on the x-axis. Points with smaller x-values will have one color, and points with larger x-values will have another, creating a gradient.

cmap='viridis': This sets the colormap to be used. 'viridis' is a perceptually uniform colormap, meaning it progresses smoothly through colors that are easy to distinguish, even for color-blind individuals. It defines how the c values (in this case, x_data) are mapped to actual colors.

s=y_data*10: This argument controls the size of each marker (point) in the scatter plot. By multiplying y_data by 10, the size of each point will be proportional to its corresponding y-value. Larger y-values will result in larger points.

alpha=0.7: This sets the transparency of the markers. alpha values range from 0 (completely transparent) to 1 (completely opaque). A value of 0.7 means the points are 70% opaque, allowing you to see overlapping points or gridlines underneath.



Histograms

## What is a Histogram?
A histogram is a graphical representation of the distribution of numerical data. It is an estimate of the probability distribution of a continuous variable. To construct a histogram, the first step is to 'bin' (or 'bucket') the range of values—that is, divide the entire range of values into a series of intervals—and then count how many values fall into each interval. The bins are usually specified as consecutive, non-overlapping intervals of a variable. The bins must be adjacent and are often (but not required to be) of equal size.

## Why are Histograms Used?
Histograms are excellent for:

*   **Visualizing Data Distribution:** They show the shape of the data's distribution, revealing whether it's normal, skewed, multimodal, etc.
*   **Identifying Central Tendency and Spread:** You can easily see where the data is centered and how spread out it is.
*   **Detecting Outliers and Gaps:** Unusual data points or missing ranges can become apparent.
*   **Comparing Distributions:** Multiple histograms can be used to compare the distributions of different datasets or subsets.

## Real-World Uses
Histograms are widely used in various fields:

*   **Statistics/Data Analysis:** To understand the underlying distribution of any quantitative variable, such as age, income, test scores, or waiting times.
*   **Quality Control:** To check the variation in product dimensions or process performance.
*   **Image Processing:** To analyze the distribution of pixel intensities in an image.
*   **Demographics:** To show the distribution of a population across different age groups.
*   **Environmental Science:** To visualize the distribution of pollution levels, temperatures, or rainfall data.

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate some random data (e.g., ages of people)
np.random.seed(42)
data = np.random.normal(loc=30, scale=10, size=1000) # Mean 30, Std Dev 10, 1000 data points

# Create the histogram
plt.figure(figsize=(10, 6))
plt.hist(data, bins=20, color='skyblue', edgecolor='black', alpha=0.7)

# Add title and labels
plt.title('Distribution of Sample Data (Histogram)')
plt.xlabel('Value')
plt.ylabel('Frequency')

# Add grid for better readability (optional)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.show()
```

## What is a Mesh Grid (`numpy.meshgrid`)?

`numpy.meshgrid` is a function in the NumPy library that creates a rectangular grid out of two 1D arrays representing the Cartesian coordinates (x and y, or more generally, any two variables). It returns two 2D arrays: one for the x-coordinates and one for the y-coordinates, where each pair `(X[i, j], Y[i, j])` corresponds to a point on the grid.

**Why is it Used?**

Mesh grids are fundamental for:

*   **Evaluating 2D functions:** When you have a function `f(x, y)` and you want to plot its values over a 2D domain, `meshgrid` provides all the `(x, y)` pairs you need.
*   **Contour Plots:** To draw contour lines of a function.
*   **Surface Plots:** To create 3D visualizations of functions.
*   **Vector Fields:** To define the points where vectors originate.

Let's demonstrate with an example: creating a 3D surface plot of a mathematical function.

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D # Required for 3D plotting

# 1. Define the range for x and y coordinates
x_coords = np.linspace(-5, 5, 50) # 50 points from -5 to 5 for x
y_coords = np.linspace(-5, 5, 50) # 50 points from -5 to 5 for y

# 2. Create the mesh grid
# X and Y will be 2D arrays, representing the x and y coordinates of each point on the grid
X, Y = np.meshgrid(x_coords, y_coords)

# 3. Define a 2D function to plot (e.g., a saddle point function)
Z = X**2 - Y**2

# 4. Create the 3D plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d') # Add a 3D subplot

# Plot the surface
ax.plot_surface(X, Y, Z, cmap='viridis')

# Add labels and title
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
ax.set_title('3D Surface Plot using Mesh Grid (Z = X^2 - Y^2)')

plt.show()
```

fig.add_subplot(...): This is a method of a Matplotlib Figure object (fig). A Figure can contain one or more Axes objects (which are the actual plots). add_subplot() is used to add a new Axes to the figure.

111: This is a shorthand for 1 row, 1 column, and the 1st subplot. So, it means you're creating a single subplot that occupies the entire figure.

Alternatively, you could write fig.add_subplot(1, 1, 1) which is more explicit.
projection='3d': This is the crucial argument that tells Matplotlib to create a 3D Axes object instead of the default 2D Axes. Without this, you wouldn't be able to plot in three dimensions.

ax = ...: The result of fig.add_subplot(...) is an Axes object, which is then assigned to the variable ax. This ax object is what you'll use to actually plot your 3D data (e.g., ax.plot_surface, ax.scatter, ax.set_xlabel, etc.).

## What is a Contour Plot?
A contour plot (sometimes called a level plot) is a graphical representation of a 3D surface `Z = f(x, y)` by plotting constant `Z` slices, called contours, on a 2D plane. It's like looking at a topographical map where lines connect points of equal elevation.

## Why are Contour Plots Used?
Contour plots are excellent for:

*   **Visualizing 3D data in 2D:** They allow you to understand the shape and features of a 3D surface without needing a full 3D rendering.
*   **Identifying gradients and critical points:** Closely spaced contour lines indicate a steep slope, while widely spaced lines indicate a gentler slope. Peaks, valleys, and saddle points are also easily identifiable.
*   **Comparing multiple functions:** You can overlay contour plots of different functions to compare their behavior.

## Real-World Uses
Contour plots are widely used in various fields:

*   **Meteorology:** Showing lines of equal atmospheric pressure (isobars) or temperature (isotherms).
*   **Geography/Cartography:** Topographical maps use contour lines to represent elevation.
*   **Engineering:** Analyzing stress distributions, fluid flow, or electromagnetic fields.
*   **Optimization:** Visualizing objective functions to find minima or maxima.
*   **Machine Learning:** Plotting decision boundaries or loss landscapes of models.

```python
import matplotlib.pyplot as plt
import numpy as np

# We'll reuse the X, Y, and Z data from the previous mesh grid example
# If these variables are not in the current kernel state, uncomment and run the following:
# x_coords = np.linspace(-5, 5, 50)
# y_coords = np.linspace(-5, 5, 50)
# X, Y = np.meshgrid(x_coords, y_coords)
# Z = X**2 - Y**2

# Create the contour plot
plt.figure(figsize=(10, 8))
# plt.contour creates contour lines
CS = plt.contour(X, Y, Z, levels=15, cmap='viridis') # levels specify the number of contour lines

# plt.contourf creates filled contours
# plt.contourf(X, Y, Z, levels=15, cmap='viridis')

# Add labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Contour Plot of Z = X^2 - Y^2')

# Add a color bar to indicate the Z values
plt.colorbar(CS, label='Z-value')

# Add labels to the contour lines (optional)
plt.clabel(CS, inline=True, fontsize=8)

plt.show()
```
