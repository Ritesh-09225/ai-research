## Introduction to Python's `random` Module

The `random` module in Python implements pseudo-random number generators for various distributions. It's used for tasks ranging from shuffling a list to simulating complex systems. While the numbers are not truly random (they are generated by a deterministic algorithm), they are sufficiently random for most purposes.

```python
import random

# Generate a random integer between 1 and 10 (inclusive)
random_number = random.randint(1, 10)
print(f"A random number between 1 and 10: {random_number}")
```

##Seeding
To ensure reproducibility when working with random numbers, you can set a 'seed' for the random number generator. This means that every time you run your code with the same seed, you'll get the exact same sequence of 'random' numbers.
<br>

In Python, you typically do this using the random.seed() function from the random module. You pass an integer value to this function as the seed. Here's an example:

```python
import random

# Set the seed for reproducibility
random.seed(42)

print(f"First random number with seed 42: {random.randint(1, 100)}")
print(f"Second random number with seed 42: {random.randint(1, 100)}")

# If you run the following block again, you'll get the same numbers
# Resetting the seed or setting a new one will generate a new sequence
random.seed(42)
print(f"First random number again with seed 42: {random.randint(1, 100)}")

random.seed(123) # Set a different seed
print(f"Random number with new seed 123: {random.randint(1, 100)}")
```

Why is this important?

Debugging: When you encounter a bug that only appears with specific random values, setting a seed allows you to reliably reproduce the scenario and debug it.

Research and Peer Review: In scientific research, it's crucial that experiments are reproducible. Setting a seed ensures that others can get the same results from your code.

Testing: When testing algorithms that involve randomness, a fixed seed ensures consistent test outcomes.

Remember that once you set a seed, the sequence of numbers generated by subsequent calls to random functions (like random.randint, random.random, etc.) will be determined by that seed until a new seed is set.

#Generate a list of random numbers

```python
import random

# Generate a list of 5 random integers between 1 and 100
random_numbers_list = [random.randint(1, 100) for _ in range(5)]
print(f"List of random integers: {random_numbers_list}")

# You can also use random.random() for floats between 0.0 and 1.0
random_floats_list = [random.random() for _ in range(5)]
print(f"List of random floats: {random_floats_list}")
```

##random.random() and random.uniform
random.random(): This function generates a random float x such that 0.0 <= x < 1.0. It does not take any arguments, and its range is always fixed to be between 0 (inclusive) and 1 (exclusive).

random.uniform(a, b): This function generates a random float x such that a <= x <= b. You specify the lower (a) and upper (b) bounds for the random number. The generated number can be equal to a or b.

##random.choice()

```python
import random

# Let's demonstrate random.choice()
# This function returns a randomly selected element from a non-empty sequence.

choices = ['apple', 'banana', 'cherry', 'date', 'elderberry']
random_choice = random.choice(choices)
print(f"Randomly chosen fruit: {random_choice}")

# Another example with numbers
numbers = [10, 20, 30, 40, 50]
random_number_choice = random.choice(numbers)
print(f"Randomly chosen number: {random_number_choice}")
```

```python
import random

def guessing_game():
    print("Welcome to the Number Guessing Game!")
    print("I'm thinking of a number between 1 and 100.")

    # Generate a random number between 1 and 100
    secret_number = random.randint(1, 100)
    attempts = 0

    while True:
        try:
            guess = int(input("Take a guess: "))
            attempts += 1

            if guess < secret_number:
                print("Your guess is too low. Try again!")
            elif guess > secret_number:
                print("Your guess is too high. Try again!")
            else:
                print(f"Congratulations! You guessed the number {secret_number} in {attempts} attempts!")
                break
        except ValueError:
            print("Invalid input. Please enter an integer.")

# Run the game
guessing_game()
```

## The `sys` Module

The `sys` module in Python provides access to system-specific parameters and functions. It allows you to interact with the Python interpreter directly and provides information about the system environment, such as the Python version, command-line arguments, and the search path for modules.

```python
import sys

print(f"Python Version: {sys.version}")
```

#sys.path
The sys.path in Python is a list of strings that specifies the search path for modules. When you import a module, Python looks for it in the directories listed in sys.path in the order they appear.

Here's a breakdown of how it works:

What it is: It's a regular Python list, which means you can inspect it, add to it, remove from it, or modify its elements just like any other list.

Its purpose: It tells the Python interpreter where to look for modules and packages you try to import. This includes standard library modules, third-party libraries, and your own custom modules.

How it's populated: When Python starts up, sys.path is initialized with:

The directory containing the input script (or the current directory if Python is run interactively).
The contents of the PYTHONPATH environment variable.

The installation-dependent default path, which includes directories for the standard library and site-packages.

Why it's important: If Python can't find a module you're trying to import, it will raise an ImportError. This usually means the module's location is not in sys.path.

Modifying sys.path: While generally not recommended for managing dependencies (package managers like pip are better), you can temporarily modify sys.path within a script to add directories where your modules are located. This is often done using sys.path.append('path/to/your/module/directory').

```python
import sys

# Print the current sys.path
print("Current sys.path:")
for p in sys.path:
    print(p)
```

## The `os` Module

The `os` module in Python provides a way of using operating system dependent functionality. It allows you to interact with the underlying operating system, performing tasks such as:

*   **Interacting with the file system:** Creating, deleting, renaming files and directories, checking file existence, getting file sizes, etc.
*   **Managing processes:** Spawning new processes, getting process IDs, etc.
*   **Environment variables:** Accessing and modifying environment variables.
*   **Path manipulation:** Joining, splitting, and normalizing paths.

Essentially, the `os` module gives your Python programs the ability to communicate with the operating system on which they are running, making them more robust and capable of handling system-level tasks.

```python
import os

# Get the current working directory
current_directory = os.getcwd()
print(f"Current working directory: {current_directory}")

# List all files and directories in the current working directory
print("Contents of the current directory:")
for item in os.listdir(current_directory):
    print(item)
```

### Creating New Directories

- `os.mkdir(path)`: Creates a single directory named `path`. It will raise an `OSError` if the directory already exists or if the parent directory does not exist.
- `os.makedirs(path, exist_ok=False)`: Creates all intermediate directories if they don't exist. If `exist_ok` is `True`, it won't raise an error if the target directory already exists.

```python
import os

# Define a directory name
new_directory = 'my_new_folder'

# Create a new directory using os.mkdir()
try:
    os.mkdir(new_directory)
    print(f"Directory '{new_directory}' created successfully.")
except FileExistsError:
    print(f"Directory '{new_directory}' already exists.")
except OSError as e:
    print(f"Error creating directory '{new_directory}': {e}")

# Create nested directories using os.makedirs()
nested_directory = 'parent_folder/child_folder/grandchild_folder'
try:
    os.makedirs(nested_directory, exist_ok=True) # exist_ok=True prevents error if it already exists
    print(f"Nested directory '{nested_directory}' created successfully.")
except OSError as e:
    print(f"Error creating nested directory '{nested_directory}': {e}")

# Verify creation by listing the current directory again
print("\nContents of the current directory after creating new folders:")
for item in os.listdir('.'):
    print(item)
```

### Removing Directories

- `os.rmdir(path)`: Removes an empty directory named `path`. It will raise an `OSError` if the directory is not empty or does not exist.
- `shutil.rmtree(path)`: Recursively deletes a directory and its contents (files and subdirectories). This is a more powerful and potentially dangerous function, so use it with caution.

```python
import os
import shutil

# Let's remove the empty directory 'my_new_folder'
try:
    os.rmdir('my_new_folder')
    print("Directory 'my_new_folder' removed successfully.")
except OSError as e:
    print(f"Error removing 'my_new_folder': {e}")

# Now, let's remove the nested directory 'parent_folder' and its contents
# We need shutil.rmtree() for non-empty directories
try:
    shutil.rmtree('parent_folder')
    print("Directory 'parent_folder' and its contents removed successfully.")
except OSError as e:
    print(f"Error removing 'parent_folder': {e}")

# Verify removal by listing the current directory again
print("\nContents of the current directory after removing folders:")
for item in os.listdir('.'):
    print(item)
```

## `os.path.join()`

The `os.path.join()` function is used to concatenate one or more path components intelligently. It inserts the appropriate directory separator (e.g., `/` on Unix-like systems, `\` on Windows) between path components, making your code platform-independent and preventing issues with hardcoded separators.

### Why use `os.path.join()`?

*   **Platform Independence**: Automatically uses the correct path separator for the operating system.
*   **Error Prevention**: Helps avoid common errors when manually concatenating strings for paths, such as missing or extra separators.
*   **Readability**: Makes path construction clearer and easier to understand.

```python
import os

# Example 1: Joining simple path components
dir_name = 'my_documents'
file_name = 'report.txt'
full_path = os.path.join(dir_name, file_name)
print(f"Full path: {full_path}")

# Example 2: Joining multiple directory components
base_dir = 'user_data'
sub_dir1 = 'photos'
sub_dir2 = 'vacation'
image_file = 'beach.jpg'
combined_path = os.path.join(base_dir, sub_dir1, sub_dir2, image_file)
print(f"Combined path: {combined_path}")

# Example 3: Handling absolute paths
# If any component is an absolute path, all previous components are discarded.
abs_path = '/home/user'
relative_path = 'data/file.csv'
result_path = os.path.join(abs_path, relative_path)
print(f"Result with absolute path: {result_path}")

# Example 4: When a component starts with a separator
# This also works like an absolute path and discards previous parts.
path_start = 'folder1'
path_mid = 'folder2'
path_end = '/new_root/file.txt'
result_path_2 = os.path.join(path_start, path_mid, path_end)
print(f"Result with component starting with separator: {result_path_2}")
```

## The `datetime` Module

The `datetime` module in Python supplies classes for working with dates and times in both simple and complex ways. It provides a more robust and flexible approach compared to just using timestamps or string manipulations.

### Key Classes:

*   **`date`**: Represents a date (year, month, day).
*   **`time`**: Represents a time (hour, minute, second, microsecond).
*   **`datetime`**: A combination of a date and a time.
*   **`timedelta`**: Represents a duration, the difference between two `date`, `time`, or `datetime` instances.
*   **`tzinfo`**: An abstract base class for time zone information objects.

### Why use `datetime`?

*   **Object-Oriented**: Dates and times are represented as objects, making them easier to manipulate.
*   **Time Zone Support**: Handles time zone conversions and daylight saving time.
*   **Arithmetic Operations**: Supports addition and subtraction of dates and times using `timedelta`.
*   **Formatting**: Easy conversion to and from strings in various formats.

```python
import datetime

# Get the current date and time
now = datetime.datetime.now()
print(f"Current date and time: {now}")

# Get the current date only
today = datetime.date.today()
print(f"Current date: {today}")

# Create a specific datetime object
christmas = datetime.datetime(2024, 12, 25, 10, 30, 0)
print(f"Christmas 2024: {christmas}")

# Access components of a datetime object
print(f"Year: {christmas.year}, Month: {christmas.month}, Day: {christmas.day}")
print(f"Hour: {christmas.hour}, Minute: {christmas.minute}, Second: {christmas.second}")

# Format a datetime object as a string
formatted_date = now.strftime("%Y-%m-%d %H:%M:%S")
print(f"Formatted current date and time: {formatted_date}")

# Calculate a timedelta (difference between two datetimes)
time_until_christmas = christmas - now
print(f"Time until Christmas: {time_until_christmas}")

# Add/subtract timedeltas
one_day = datetime.timedelta(days=1)
tomorrow = now + one_day
print(f"Tomorrow: {tomorrow}")
```

## The `math` Module

The `math` module provides access to common mathematical functions and constants. It's built-in, so you don't need to install it, just import it.

## The `collections` Module

The `collections` module in Python implements specialized container datatypes providing alternatives to Python's general purpose built-in containers `dict`, `list`, `set`, and `tuple`.

### Why use `collections`?

*   **Efficiency**: Some `collections` types are optimized for specific use cases, offering better performance than their generic counterparts.
*   **Convenience**: They provide ready-to-use functionalities for common tasks, reducing the amount of boilerplate code you need to write.
*   **Readability**: Using the right specialized container can make your code clearer and easier to understand.

Let's look at some of the most useful data structures provided by the `collections` module.

### `collections.Counter`

A `Counter` is a `dict` subclass for counting hashable objects. It's an unordered collection where elements are stored as dictionary keys and their counts are stored as dictionary values. Counts are allowed to be any integer value including zero or negative counts.

```python
from collections import Counter

# Count the occurrences of characters in a string
word = "banana"
char_counts = Counter(word)
print(f"Character counts for '{word}': {char_counts}")

# Count occurrences in a list
fruits = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']
fruit_counts = Counter(fruits)
print(f"Fruit counts: {fruit_counts}")

# Accessing counts
print(f"Count of 'apple': {fruit_counts['apple']}")
print(f"Count of 'grape' (non-existent): {fruit_counts['grape']}") # Returns 0 for non-existent items

# Most common elements
print(f"Two most common fruits: {fruit_counts.most_common(2)}")

# Adding counts from another Counter or iterable
another_basket = ['kiwi', 'apple', 'banana']
fruit_counts.update(another_basket)
print(f"Fruit counts after update: {fruit_counts}")
```

### `collections.defaultdict`

A `defaultdict` is a subclass of `dict` that calls a factory function to supply missing values. This is useful when you want a dictionary to automatically create default values for keys that don't exist yet, instead of raising a `KeyError`.

```python
from collections import defaultdict

# Using defaultdict with a list as the default factory
# This is great for grouping items
s_list = [('yellow', 1), ('blue', 2), ('yellow', 3), ('blue', 4), ('red', 1)]
d_list = defaultdict(list)
for k, v in s_list:
    d_list[k].append(v)
print(f"Grouped list items: {d_list}")

# Using defaultdict with an int as the default factory (for counting)
# Similar to Counter, but demonstrates the concept
sentence = "the quick brown fox jumps over the lazy dog"
word_counts = defaultdict(int)
for word in sentence.split():
    word_counts[word] += 1
print(f"Word counts (defaultdict): {word_counts}")

# Using defaultdict with a set as the default factory
# For storing unique items per key
maps = [('A', 1), ('B', 2), ('A', 3), ('C', 2)]
d_set = defaultdict(set)
for k, v in maps:
    d_set[k].add(v)
print(f"Grouped unique items: {d_set}")

# Accessing a non-existent key will create it with the default value
print(f"Accessing a non-existent key 'D': {d_set['D']}")
print(f"Dictionary after accessing 'D': {d_set}")
```

### `collections.deque` (Double-ended Queue)

A `deque` (double-ended queue) is a list-like container with fast appends and pops from either end. While Python's built-in `list` also supports `append()` and `pop()` operations, `deque` provides better performance for adding or removing elements from the beginning of the sequence.

```python
from collections import deque

# Create a deque
d = deque(['a', 'b', 'c'])
print(f"Initial deque: {d}")

# Append elements to the right
d.append('d')
print(f"After append('d'): {d}")

# Append elements to the left
d.appendleft('z')
print(f"After appendleft('z'): {d}")

# Pop elements from the right
right_popped = d.pop()
print(f"Popped from right: {right_popped}, deque now: {d}")

# Pop elements from the left
left_popped = d.popleft()
print(f"Popped from left: {left_popped}, deque now: {d}")

# Extend the deque from an iterable
d.extend(['e', 'f', 'g'])
print(f"After extend(['e', 'f', 'g']): {d}")

# Extend the deque from the left
d.extendleft(['x', 'y'])
print(f"After extendleft(['x', 'y']): {d}")

# Rotate the deque
d.rotate(1) # Rotate right by 1
print(f"After rotate(1): {d}")
d.rotate(-2) # Rotate left by 2
print(f"After rotate(-2): {d}")

# Deques can also have a maximum length
limited_deque = deque(maxlen=3)
limited_deque.append(1)
limited_deque.append(2)
limited_deque.append(3)
print(f"Limited deque (full): {limited_deque}")
limited_deque.append(4) # Appending 4 pushes out 1 from the left
print(f"Limited deque after append(4): {limited_deque}")
```
